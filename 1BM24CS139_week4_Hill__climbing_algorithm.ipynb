```python
import random

def get_conflicts(state):
    """Calculate the number of attacking pairs of queens (heuristic cost)."""
    conflicts = 0
    n = len(state)
    for i in range(n):
        for j in range(i + 1, n):
            # Check if they are in the same column
            if state[i] == state[j]:
                conflicts += 1
            # Check diagonals
            elif abs(state[i] - state[j]) == abs(i - j):
                conflicts += 1
    return conflicts

def get_neighbors(state):
    """Generate all possible neighbor states by moving one queen in her row."""
    neighbors = []
    n = len(state)
    for row in range(n):
        for col in range(n):
            if state[row] != col:
                neighbor = list(state)
                neighbor[row] = col
                neighbors.append(neighbor)
    return neighbors

def hill_climbing_4_queens():
    """Hill climbing algorithm to solve the 4-Queens problem."""
    # 1. Start with a random initial state
    current_state = [random.randint(0, 3) for _ in range(4)]
    current_conflicts = get_conflicts(current_state)

    print(f"Initial random state: {current_state} (Conflicts: {current_conflicts})")

    step = 0
    while True:
        neighbors = get_neighbors(current_state)
        if not neighbors:
            break

        # Find the neighbor(s) with the minimum conflicts
        neighbor_evals = [(get_conflicts(n), n) for n in neighbors]
        best_neighbor_conflicts, best_neighbor = min(neighbor_evals, key=lambda x: x[0])

        # If the best neighbor is not better than current, we've hit a local optimum / peak
        if best_neighbor_conflicts >= current_conflicts:
            break

        current_state = best_neighbor
        current_conflicts = best_neighbor_conflicts
        step += 1
        print(f"Step {step}: {current_state} (Conflicts: {current_conflicts})")

    return current_state, current_conflicts

# Run the algorithm
solution, conflicts = hill_climbing_4_queens()

if conflicts == 0:
    print(f"\nSuccess! Found a solution: {solution}")
else:
    print(f"\nStuck in a local optimum at state: {solution} with {conflicts} conflicts.")
```

Note: Standard Hill Climbing can easily get stuck in a local minimum (where conflicts > 0 but no single move improves it). If the run above gets stuck, you can simply execute the cell again to start from a new random initial state.
