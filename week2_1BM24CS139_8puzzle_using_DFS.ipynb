```python
class PuzzleState:
    def __init__(self, board, parent=None, move=None):
        self.board = board  # Tuple of tuples representing the 3x3 board
        self.parent = parent
        self.move = move
        self.empty_pos = self.find_empty()

    def __eq__(self, other):
        return self.board == other.board

    def __hash__(self):
        return hash(self.board)

    def find_empty(self):
        for r in range(3):
            for c in range(3):
                if self.board[r][c] == 0:
                    return (r, c)
        return None

    def get_neighbors(self):
        neighbors = []
        er, ec = self.empty_pos
        moves = [(0, 1), (0, -1), (1, 0), (-1, 0)]  # Right, Left, Down, Up

        for dr, dc in moves:
            nr, nc = er + dr, ec + dc

            if 0 <= nr < 3 and 0 <= nc < 3:
                new_board_list = [list(row) for row in self.board]
                new_board_list[er][ec], new_board_list[nr][nc] = new_board_list[nr][nc], new_board_list[er][ec]
                new_board_tuple = tuple(tuple(row) for row in new_board_list)
                neighbors.append(PuzzleState(new_board_tuple, self, (nr, nc)))
        return neighbors

    def display(self):
        for row in self.board:
            print(" ".join(map(str, row)))
        print()

def dfs(initial_state, goal_state, depth_limit=20):
    stack = [(initial_state, 0)]  # (state, current_depth)
    visited = {initial_state}

    while stack:
        current_state, current_depth = stack.pop()

        if current_state == goal_state:
            return current_state

        if current_depth < depth_limit:
            for neighbor in reversed(current_state.get_neighbors()): # Reverse to explore left-most child first
                if neighbor not in visited:
                    visited.add(neighbor)
                    stack.append((neighbor, current_depth + 1))
    return None

def get_path(solution_state):
    path = []
    current = solution_state
    while current:
        path.append(current)
        current = current.parent
    return path[::-1]

# Define initial and goal states
initial_board = (
    (2, 8, 3),
    (1, 6, 4),
    (7, 0, 5)
)

goal_board = (
    (1, 2, 3),
    (8, 0, 4),
    (7, 6, 5)
)

initial_state = PuzzleState(initial_board)
goal_state = PuzzleState(goal_board)

print("Initial State:")
initial_state.display()

print("Goal State:")
goal_state.display()

print("Solving with DFS...")
solution = dfs(initial_state, goal_state, depth_limit=50) # Increased depth limit for more complex cases

if solution:
    print("Solution Found!")
    path = get_path(solution)
    for i, state in enumerate(path):
        print(f"Step {i}:")
        state.display()
else:
    print("No solution found within the given depth limit.")
```
