```python
import math
import random

def simulated_annealing_nqueens(n, initial_temp=100.0, cooling_rate=0.99, max_iterations=5000):
    """
    Solves the N-Queens problem using Simulated Annealing.
    The state is represented as a list of length N, where the index is the row
    and the value is the column position of the queen in that row.
    """

    # --- Helper 1: Cost Function ---
    # Counts how many pairs of queens are attacking each other
    def get_cost(state):
        conflicts = 0
        for i in range(n):
            for j in range(i + 1, n):
                # Check if they share the same column or same diagonal
                if state[i] == state[j] or abs(state[i] - state[j]) == abs(i - j):
                    conflicts += 1
        return conflicts

    # --- Helper 2: Neighbor Generator ---
    # Generates a neighbor by moving one random queen to a different random column
    def get_neighbor(state):
        neighbor = list(state)
        row = random.randint(0, n - 1)
        new_col = random.randint(0, n - 1)

        # Ensure we actually pick a different column
        while new_col == state[row]:
            new_col = random.randint(0, n - 1)

        neighbor[row] = new_col
        return neighbor

    # --- Helper 3: Board Printer ---
    def print_board(state):
        for row in range(n):
            line = ["Q" if state[row] == col else "." for col in range(n)]
            print(" ".join(line))

    # Line 1: current <- initial state (randomly place one queen per row)
    current = [random.randint(0, n - 1) for _ in range(n)]
    current_cost = get_cost(current)

    # Line 2: T <- a large positive value
    T = initial_temp

    iteration = 0

    # Line 3: while T > 0 do (Also stop if a perfect solution is found)
    while T > 0.0001 and current_cost > 0 and iteration < max_iterations:
        iteration += 1

        # Line 4: next <- a random neighbour of current
        next_state = get_neighbor(current)
        next_cost = get_cost(next_state)

        # Line 5: ΔE <- current.cost - next.cost
        delta_E = current_cost - next_cost

        # Line 6: if ΔE > 0 then
        if delta_E > 0:
            # Line 7: current <- next
            current = next_state
            current_cost = next_cost

        # Line 8: else
        else:
            # Line 9: current <- next with probability p = e^(ΔE / T)
            p = math.exp(delta_E / T)
            if random.random() < p:
                current = next_state
                current_cost = next_cost

        # Line 10: end if

        # Line 11: decrease T
        T = T * cooling_rate

    # Line 12: end while

    # Display Results
    print(f"--- N-Queens Results (N = {n}) ---")
    print(f"Total Iterations: {iteration}")
    print(f"Final Conflicts (Cost): {current_cost}")
    print(f"Final State (Columns per row): {current}\n")

    if current_cost == 0:
        print("Success! Found a valid board configuration:")
        print_board(current)
    else:
        print("Could not find a perfect solution within the temperature limit.")
        print("Final board arrangement:")
        print_board(current)

    # Line 13: return current
    return current

# Execute for an 8-Queens problem
if __name__ == "__main__":
    # You can change 8 to other sizes like 4, 5, or 10
    simulated_annealing_nqueens(n=8, initial_temp=50.0, cooling_rate=0.98)
```
