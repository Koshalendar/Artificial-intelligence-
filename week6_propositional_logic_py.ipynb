```python
NOT = "NOT"
AND = "AND"
OR = "OR"
IMPLIES = "IMPLIES"

def tt_entails(kb, alpha):
    symbols = list(get_symbols(kb) | get_symbols(alpha))
    return tt_check_all(kb, alpha, symbols, {})

def tt_check_all(kb, alpha, symbols, model):
    if not symbols:
        if pl_true(kb, model):
            return pl_true(alpha, model)
        else:
            return True
    else:
        P = symbols[0]
        rest = symbols[1:]

        model_true = model.copy()
        model_true[P] = True

        model_false = model.copy()
        model_false[P] = False

        return (tt_check_all(kb, alpha, rest, model_true) and
                tt_check_all(kb, alpha, rest, model_false))

def get_symbols(expr):
    if isinstance(expr, str):
        return {expr} if expr not in {NOT, AND, OR, IMPLIES} else set()
    elif isinstance(expr, tuple):
        symbols = set()
        for element in expr:
            symbols.update(get_symbols(element))
        return symbols
    return set()

def pl_true(expr, model):
    if isinstance(expr, str):
        return model[expr]

    if isinstance(expr, tuple):
        if expr[0] == NOT:
            return not pl_true(expr[1], model)

        left, op, right = expr[0], expr[1], expr[2]
        if op == AND:
            return pl_true(left, model) and pl_true(right, model)
        elif op == OR:
            return pl_true(left, model) or pl_true(right, model)
        elif op == IMPLIES:
            return (not pl_true(left, model)) or pl_true(right, model)

    raise ValueError(f"Unknown expression format: {expr}")

if __name__ == "__main__":
    print("--- Truth-Table Model Checking Algorithm ---")
    print("Format examples:")
    print("  Single symbol: \"Rain\"")
    print("  Expression: (\"Rain\", IMPLIES, \"Wet\")")
    print("  Complex KB: ((\"Rain\", IMPLIES, \"Wet\"), AND, \"Rain\")\n")

    try:
        kb_input = input("Enter your Knowledge Base (KB): ")
        alpha_input = input("Enter your Query (alpha): ")

        user_kb = eval(kb_input)
        user_alpha = eval(alpha_input)

        result = tt_entails(user_kb, user_alpha)
        print(f"\nDoes KB entail Alpha? {result}")
    except Exception as e:
        print(f"\nError parsing input: {e}")
        print("Make sure to use correct tuple formatting, quotes for symbols, and capitalized operators.")
```
