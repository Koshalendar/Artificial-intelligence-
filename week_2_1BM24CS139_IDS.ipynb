```python
def print_puzzle(state):
    print(state[0], state[1], state[2])
    print(state[3], state[4], state[5])
    print(state[6], state[7], state[8])
    print()

def dfs(state, goal, depth,limit,visited, path):

    # Goal reached
    if state == goal:
        path.append(state)
        return True

    if depth == limit:
      return False

    # Mark state as visited
    # Note: For IDDFS, 'visited' should typically be reset for each DLS call
    # or modified to handle states revisited at different depths. For simpler
    # puzzles, a simple visited set for the current DLS iteration might suffice.
    # However, to strictly follow the IDDFS concept of re-exploring, 'visited' should be passed empty
    # or local to the DLS call within IDDFS. Here, I'll pass an empty one in iddfs.

    path.append(state)

    # Find blank space
    blank = state.index(0)

    row = blank // 3
    col = blank % 3

    # Up, Down, Left, Right
    moves = [
        (-1, 0),
        (1, 0),
        (0, -1),
        (0, 1)
    ]

    for dr, dc in moves:

        new_row = row + dr
        new_col = col + dc

        # Check valid position
        if 0 <= new_row < 3 and 0 <= new_col < 3:

            new_blank = new_row * 3 + new_col

            # Create new state
            new_state = list(state)

            # Swap blank with tile
            new_state[blank], new_state[new_blank] = \
                new_state[new_blank], new_state[blank]

            new_state = tuple(new_state)

            # DFS on new state
            if new_state not in visited:
                visited.add(new_state) # Add to visited for THIS DLS call
                if dfs(new_state, goal,depth+1,limit, visited, path):
                    return True

    # Backtracking
    path.pop()

    return False


def iddfs(start_state, goal_state, max_limit=30):
    """Iterative Deepening Depth-First Search"""
    for limit in range(max_limit + 1):
        visited = set() # Visited set must be reset for each depth limit
        path = []       # Path must be reset for each depth limit
        depth = 0

        print(f"Trying with depth limit = {limit}")

        if dfs(start_state, goal_state, depth, limit, visited, path):
            return path
    return None # No solution found within max_limit


# Starting state
start = (
    2,8,3,
    1,6,4,
    7,0,5
)

# Goal state
goal = (
    1, 2, 3,
    8,0,4,
    7,6,5
)

# Run IDDFS
solution_path = iddfs(start, goal)

if solution_path:

    print("Solution found!")
    print()

    for i, state in enumerate(solution_path):

        print("Step", i)
        print_puzzle(state)

else:
    print("No solution found within the specified limits.")

# Removed incorrect print statements regarding complexity
```
